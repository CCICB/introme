```python
df = preprocess(train_mode=True, df=df)
```

```python
for col in df.columns:
    if col == "REF" or col == "ALT":
        continue
    # if not pd.api.types.is_numeric_dtype(df[col]):
    print(f"Column: {col}")
    print("Unique values:", df[col].unique())
    print()

for col in df.columns:
    # Find non-numeric values/NaN/None and print them
    non_numeric_mask = ~pd.to_numeric(df[col], errors='coerce').notna()
    if non_numeric_mask.any():
        non_numeric_values = df.loc[non_numeric_mask, col].unique()
        print(f"Non-numeric values in column '{col}': {non_numeric_values}")
        print()
```
